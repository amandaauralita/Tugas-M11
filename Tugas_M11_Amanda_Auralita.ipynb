```python
#NOMER 3.1
def hitung_luas_persegi(sisi):
  luas_persegi = sisi*sisi
  print(f"Luas persegi adalah: {luas_persegi}")
  return luas_persegi
def rata_rata(luas_persegi1,luas_persegi2,luas_persegi3):
  rata_rata = (luas_persegi1 + luas_persegi2 + luas_persegi3) / 3
  return rata_rata
luas_persegi1 = hitung_luas_persegi(5)
luas_persegi2 = hitung_luas_persegi(10)
luas_persegi3 = hitung_luas_persegi(15)
print(f"Rata-rata luas persegi dari ketiga luas adalah: {rata_rata(luas_persegi1,luas_persegi2,luas_persegi3)}")
```
